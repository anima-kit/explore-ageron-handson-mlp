```python
import base64
import requests

def save_mermaid_svg(graph, filename="diagram.png"):
    # Encode diagram for URL
    graph_bytes = graph.encode("utf8")
    base64_string = base64.urlsafe_b64encode(graph_bytes).decode("ascii")
    
    # Fetch SVG from mermaid.ink
    url = f"https://mermaid.ink/svg/{base64_string}"
    response = requests.get(url)
    
    # Save to file
    with open(filename, "wb") as f:
        f.write(response.content)
    print(f"Saved to {filename}")
```

```python
ai_ml_dl_graph = """
graph LR
    ai["<b><u>Artificial Intelligence (AI)</u></b>
    Artifical systems that seem intelligent"]
    class_ai["<b><u>Classical AI</u></b>
    Artifical systems that seem intelligent but don't learn from data"]
    ml["<b><u>Machine Learning (ML)</u></b>
    Artificial systems that seem intelligent and learn from data"]
    class_ml["<b><u>Classical ML</u></b>
    Artificial systems that seem intelligent and learn from data without utilizing NNs"]
    nn["<b><u>Neural Network (NN)</u></b>
    A matrix of layers that transform data as it's passed from the input layer to each hidden layer to the output layer"]
    shallow_nn["<b><u>Shallow NN</u></b>
    A NN with less than two hidden layers"]
    dnn["<b><u>Deep NN (DNN)</u></b>
    A NN with two or more hidden layers"]
    dl["<b><u>Deep Learning (DL)</u></b>
    Artificial systems that seem intelligent and learn from data by utilizing DNNs"]

    ai --> class_ai
    ai --> ml
    ml --> class_ml
    ml --> nn
    nn --> shallow_nn
    nn --> dnn
    dnn --> dl

    classDef ai_def fill:#FFDA8A,stroke:#E69F00,color:#000
    classDef class_ai_def fill:#FAF6BD,stroke:#F0E442,color:#000
    classDef ml_def fill:#C0E3F7,stroke:#56B4E9,color:#000
    classDef class_ml_def fill:#FFD7B8,stroke:#D55E00,color:#000
    classDef nn_def fill:#B8FFEC,stroke:#009E73,color:#000
    classDef shallow_nn_def fill:#8AD4FF,stroke:#0072B2,color:#000
    classDef dnn_def fill:#DADADA,stroke:#000000,color:#000
    classDef dl_def fill:#EBCBDD,stroke:#CC79A7,color:#000

    class ai ai_def
    class class_ai class_ai_def
    class ml ml_def
    class class_ml class_ml_def
    class nn nn_def
    class shallow_nn shallow_nn_def
    class dnn dnn_def
    class dl dl_def
"""

save_mermaid_svg(ai_ml_dl_graph, 'images/mermaid/ai_vs_ml_vs_dl.svg')
```

```python
ml_types_graph = """
graph TD
    data_type["<b><u>Data Type</u></b>
    What does the training data look like?"]
    train_type["<b><u>Training Type</u></b>
    What does the training algorithm look like?"]
    pred_type["<b><u>Prediction Type</u></b>
    What does prediction on new data look like?"]
    other["<b><u>Other Types</u></b>
    Some other possible categories"]

    classDef data_type_def fill:#FFDA8A,stroke:#E69F00,color:#000
    classDef train_type_def fill:#B8FFEC,stroke:#009E73,color:#000
    classDef pred_type_def fill:#C0E3F7,stroke:#56B4E9,color:#000
    classDef other_def fill:#EBCBDD,stroke:#CC79A7,color:#000

    class data_type data_type_def
    class train_type train_type_def
    class pred_type pred_type_def
    class other other_def
"""

save_mermaid_svg(ml_types_graph, 'images/mermaid/ml_types_graph.svg')
```

```python
data_types_graph = """
graph TD
    data_type["<b><u>Data Type</u></b>
    What does the training data look like?"]
    supervised["<b><u>Supervised Learning</u></b>
    Each data point has a 'true' target"]
    unsupervised["<b><u>Unsupervised Learning</u></b>
    No data points have a 'true' target"]
    semi_supervised["<b><u>Semi-Supervised Learning</u></b>
    Some data points have a 'true' target and some don't"]
    self_supervised["<b><u>Self-Supervised Learning</u></b>
    A 'true' target can be generated for each data point"]
    reinforcement["<b><u>Reinforcement Learning</u></b>
    Data is a set of possible actions in an environment"]

    data_type --> supervised
    data_type --> unsupervised
    data_type --> semi_supervised
    data_type --> self_supervised
    data_type --> reinforcement

    classDef data_type_def fill:#FFDA8A,stroke:#E69F00,color:#000
    classDef supervised_def fill:#B8FFEC,stroke:#009E73,color:#000
    classDef unsupervised_def fill:#C0E3F7,stroke:#56B4E9,color:#000
    classDef semi_supervised_def fill:#EBCBDD,stroke:#CC79A7,color:#000
    classDef self_supervised_def fill:#FAF6BD,stroke:#F0E442,color:#000
    classDef reinforcement_def fill:#FFD7B8,stroke:#D55E00,color:#000

    class data_type data_type_def
    class supervised supervised_def
    class unsupervised unsupervised_def
    class semi_supervised semi_supervised_def
    class self_supervised self_supervised_def
    class reinforcement reinforcement_def
"""

save_mermaid_svg(data_types_graph, 'images/mermaid/data_types_graph.svg')
```
