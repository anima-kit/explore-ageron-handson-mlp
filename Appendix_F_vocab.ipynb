<span id="ai-vs-ml-vs-dl"><h1 style="text-align:center;">AI vs ML vs DL</h1></span>

<hr style="width:50%;">

![AI vs ML vs DL](./images/mermaid/ai_vs_ml_vs_dl.svg)

<span id="ai"><b><u>Artificial Intelligence (AI)</u></b></span>: An artifically created system that *seems* intelligent. This is a huge category of systems that includes both [machine learning](#ml) and [deep learning](#dl).

<span id="ml"><b><u>Machine Learning (ML)</u></b></span>: An artifically created system that learns from data. This is a more narrow category of systems than [AI](#ai) because not all [AI](#ai) *learns* from data. For example, a system that has a set of programmed rules that can be used to make decisions isn't ML, though it does appear intelligent. A system that *learns* how to predict the best decisions from input data is ML. There are many different [types of ML](#ml-types).

<span id="nn"><b><u>Neural Network (NN)</u></b></span>: A mathematical construct that loosely models how a network of neurons work in a brain. An NN is a matrix with an *input* layer, an *output* layer, and some number of *hidden* layers. Data passes through each layer of a NN until it's finally output in some transformed fashion at the output layer.

<span id="shallow-nn"><b><u>Shallow NN</u></b></span>: An [NN](#NN) with less than two hidden layers.

<span id="dnn"><b><u>Deep NN (DNN)</u></b></span>: An [NN](#nn) with more than two hidden layers. The definition of *deep* typically means two or more *hidden* layers, but this can change depending on who you ask. 

<span id="dl"><b><u>Deep Learning (DL)</u></b></span>: An artifically created system that learns from data by utilizing [DNNs](#dnn). This is an even more narrow category of [AI](#ai) and a subset of [ML](#ml) because the system still learns from data, just with a particular infrastructure. 

<span id="classical-ai"><b><u>Classical AI</u></b></span>: An artificially created system that seems intelligent but doesn't learn from data. This is a subset of [AI](#ai) that doesn't include either [ML](#ml) or [DL](#dl).

<span id="classical-ml"><b><u>Classical ML</u></b></span>: An artificially created system that learns from data without utilizing NNs. This is a subset of [ML](#ml) that doesn't include [DL](#dl).

<span id="training-basics"><h2 style="text-align:center;">Training Basics</h2></span>

<hr style="width:25%;">

<span id="training"><b><u>Training</u></b></span>: The process by which an [ML](#ml) system learns. The training process typically consists of (1) giving the system data, (2) having the system learn from the data in some way, (3) evaluating the system, (4) having the system continue learning from the data based on the evaluation, (5) repeating steps (3)-(4) until training is complete. The types of data that can be input into the system are vast and depend on the type of problem that we want our system to solve. [ML](#ml) systems can then be categorized by (amongst others) the type of data that's input for training and the way that the [model](#model) is trained (we discuss these categorizations in the [ML Types](#ml-types) section).

<span id="train-set"><b><u>Train Set</u></b></span>: (*AKA Training Set*) The data that's input into an [ML](#ml) system in order for it to learn during [training](#training). The system *[trains](#training)* on this data in order to perform some task on some other data that it hasn't seen yet (that isn't in the train set). For example, if we want to build a system that can correctly identify the type of an animal from a photo, we could give it a training set that consists of [labeled](#label) photos of animals. The system will then learn what images of dogs look like, what images of cats look like. Then, when it's given a new photo it hasn't seen, hopefully it'll be able to tell what type of animal it is.

<span id="train-instance"><b><u>Train Instance</u></b></span>: (*AKA Training Instance*) One sample of the [training set](#train-set). So, in our example above, it would be one image with its correct [label](#label). If it's an image of a dog, the training instance could be a matrix of 256x256 pixels and a [label](#label) vector of size N (for a total of N possible animal types).
$$
X = 
\begin{pmatrix} 
x_{0,0} & \cdots & x_{0,256} \\ 
\vdots &  & \vdots \\ 
x_{256,0} & \cdots & x_{256,256} 
\end{pmatrix}
\qquad
y = 
\begin{pmatrix} 
1 \\ 0 \\ \vdots \\ 0 
\end{pmatrix}
$$
where the first index of the [label](#label), $y$, corresponds to "dog".

<span id="algorithm"><b><u>Algorithm</u></b></span>: The procedure used to create a [model](#model). This typically includes picking a [model](#model) to represent our data and [training](#training) the [model](#model) on our [training set](#train-set) to produce the optimal [model parameters](#model-parameter). 

For an example of how a model could work in practice, see the definitions for [linear regression](#linear-regression) and [gradient descent](#gradient-descent).

<span id="hyperparameter"><b><u>Hyperparameter</u></b></span>: A parameter particular to the type of [training](#training) algorithm used. These aren't typically updated during [training](#training) like [model parameters](#model-parameters) are, but are usually held constant throughout the [training](#training) process. Sometimes, we might want to use a technique to find the best hyperparameters for the particular job. 

For an example of how hyperparameters work in practice, see the definition for [gradient descent](#gradient-descent). For an example of a particular hyperparameter, see the definition for [learning rate](#learning-rate).

<span id="data-basics"><h2 style="text-align:center;">Data Basics</h2></span>

<hr style="width:25%;">

<span id="label"><b><u>Label</u></b></span>: (*AKA Target*) The true value for a given data point and the variable that we want our [ML](#ml) system to predict. Let's say we want to build a system that can correctly identify the type of animal from a photo. The label in this case (the thing we want our system to predict) is the type of animal. We could then feed the system a set of images, each with their own particular label (i.e. all dog images are labeled "dog", all cat images are labeled "cat", and so forth) during [training](#training) in order to build a system that can predict the correct label (type of animal) from photos.

*Label* is often used interchangebly with *target*. However, label is typically used when the true values for our data points are discrete (like types of animals, or yes or no answers). Whereas target is typically used when the true values are continuous (like house prices or energy consumption).

We typically use the variable $y$ to denote the label (or target). When the target corresponds to a continuous value, it's just some number (e.g. the house price for this data point is 250K) and hence $y = \text{some number}$. When the label is a discrete value, it's a vector that has a 1 for the index that corresponds to the true value and 0 for all other indices. For example, let's say we want to identify animals from photos and we have three different animals to choose from ("dog", "cat", "elephant") - the system can only pick from these three for any given photo. The labels for an image of a dog, a cat, and an elephant would look like:

$$
y_\text{dog} =
\begin{pmatrix}
1 \\ 0 \\ 0
\end{pmatrix}
\qquad
y_\text{cat} =
\begin{pmatrix}
0 \\ 1 \\ 0
\end{pmatrix}
\qquad
y_\text{elephant} =
\begin{pmatrix}
0 \\ 0 \\ 1
\end{pmatrix}
$$
where the first index of the label vector denotes "dog", the second index denotes "cat", and the third index denotes "elephant".

<span id="feature"><b><u>Feature</u></b></span>: (*AKA Predictor, Attribute*) The part of the data that's actually input into the [ML](#ml) system. This is what the system learns from. For example, let's say we want to predict house prices. We could get a dataset that's full of information like house size, number of rooms, number of pools, year built, occupant names, house color, realtor name, house material, appraisal, and so forth. All of these are potential *features*. However, some of these features could be relevant to the house price (the [target](#label)), while some of them could be useless and just act to make our dataset noisy (i.e. *feature noise*).

<span id="feature-engineering"><b><u>Feature Engineering</u></b></span>: Finding the optimal set of [features](#feature) in order to [train](#training) a [model](#model) that performs well. This is definitely an art). When we build [ML](#ML) systems, we have to engineer our [features](#feature) manually. We have to pick which [features](#feature) (i.e. [feature selection](#feature-selection), or combination of [features](#feature) (i.e. [feature extraction](#feature-extraction), we think would be the most relevant to predicting our [target](#target). However, when we build [DL](#dl) systems, the optimal [features](#feature) are automatically learned (this doesn't mean that these are necessarily the optimal [features](#feature), just that the [DL](#dl) system sees them as optimal).

<span id="feature-selection"><b><u>Feature Selection</u></b></span>: Picking a set of *existing* [features](#feature) from the entire set.

<span id="feature-extraction"><b><u>Feature Extraction</u></b></span>: Creating *new* [features](#features) by combining already *existing* [features](#feature).

<span id="model-basics"><h2 style="text-align:center;">Model Basics</h2></span>

<hr style="width:25%;">

<span id="model"><b><u>Model</u></b></span>: The part of the [ML](#ml) system that *learns* from the [training set](#train-set) and is then used to perform a specific task on new data. Models have parameters (called [model parameters](#model-parameter)) which are typically changed during [training](#training) in order to find the best values to perform the task optimally. 

For an example of how a model could work in practice, see the definitions for [linear regression](#linear-regression) and [gradient descent](#gradient-descent).

<span id="model-parameter"><b><u>Model Parameter</u></b></span>: A parameter of a [model](#model) which is updated during [training](#training).

For an example of how model parameters work in practice, see the definitions for [linear regression](#linear-regression) and [gradient descent](#gradient-descent).

<span id="dof"><b><u>Degrees of Freedom</u></b></span>: A meausre of how many [model parameters](#model-parameters) can be changed in order to fit a [model](#model) to data. For a linear [model](#model) [trained](#training) to predict a house price from two [features](#features), the number of degrees of freedom is three, one intercept and one slope parameter for each [feature](#feature).

<span id="evaluation-basics"><h2 style="text-align:center;">Evaluation Basics</h2></span>

<hr style="width:25%;">

<span id="testing-basics"><h3 style="text-align:center;">Testing Basics</h3></span>

<hr style="width:10%;">

<span id="generalization"><b><u>Generalization</u></b></span>: How well a system can perform when given new data that it hasn't seen before.

<span id="generalization-error"><b><u>Generalization Error</u></b></span>: (*AKA Out-of-Sample Error*) The rate of error when [testing](#testing) your [model](#model) on cases it hasn't seen yet (e.g. how well the [model](#model) generalizes to data in the [test set](#test-set)).

<span id="representative"><b><u>Representative</u></b></span>: When your data (i.e. [training data](#train-set), [test data](#test-set), [validation data](#validation-set)) accurately depicts the type of data that you want your [model](#model) to be able to [generalize](#generalization) to. For example, if you want the [model](#model) to be able to generate summaries from medical data but your [training set](#train-set) only includes science-fiction novels, your [model](#model) will probably output suboptimal summaries. 

<span id="testing"><b><u>Testing</u></b></span>: How we evaluate the performance of our [model](#model) on new data that it hasn't seen during [training](#training). 

<span id="test-set"><b><u>Test Set</u></b></span>: (*AKA Testing Set*) The data that's used to [test](#testing) the [model](#model) after it's done [training](#training) in order to see how well the model *[generalizes](#generalization)* to new data. We need to make sure the [model](#model) doesn't see any part of the testing set during [training](#training). If it does, the [model](#model) can then optimize [model parameters](#model-parameter) with respect to the test data and our [generalization](#generalization) tests will be overly optimistic. Then, when we deploy our [model](#model) on real-world data, it will typically perform worse than we expect.

A common way of creating a test set is by splitting the [training set](#train-set) into two separate sets, one for [training](#training) and one for [testing](#testing). The percentage of data that should go into the test set isn't a fixed value but depends on how much data we need to be representative of the real-world. This can be difficult if we only take the test set as a percentage of the [training set](#train-set). For example, let's say we have Electronic Health Record (EHR) data from one particular hospital. If we build a [model](#model) that will only be used at that hospital, maybe using only that data for both [training](#training) and [testing](#testing) won't be so bad. But, if we try to use the [model](#model) at other hospitals, we might find that it performs poorly because it hasn't seen data from that hospital. It would be even better if we could use data from other hospitals to [test](#testing) our [model](#model) so that we can see how well it [generalizes](#generalization) to new data. 

<span id="performance-measure"><b><u>Performance Measure</u></b></span>: How we evaluate our [model's](#model) performance on a given task. This is typically done both while [training](#training), [validating](#holdout-validation), and during [testing](#testing). There are many types of performance measures which we discuss in more detail below.

<span id="validation-basics"><h3 style="text-align:center;">Validation Basics</h3></span>

<hr style="width:10%;">

<span id="holdout-validation"><b><u>Holdout Validation</u></b></span>: Holding out a small part of the [training set](#train-set) during [training](#training) in order to validate the [model](#model) performance when using a particular set of [hyperparameters](#hyperparameter) or a particular [model](#model). This is typically done for a number of different possible [hyperparameter](#hyperparameter) values or [models](#models), then the best performing [model](#model) is [retrained](#training) on the full [training set](#train-set) using the optimal [hyperparameters](#hyperparameter). We do this instead of [testing](#testing) our [model](#model) performance on the [test set](#test-set) because doing so would effectively mean that our [model](#model) sees the [test set](#test-set) when [training](#training) and our [generalization error](#generalization-error) may be overly optimistic (the whole reason we needed to separate our [test set](#test-set) from the [training-set](#train-set) in the first place!). 

<span id="validation-set"><b><u>Validation Set</u></b></span>:  A small subset of the [training set](#train-set) held out during [holdout validation](#holdout-validation). This is the set of data that the [model](#model) performance is evaluated on in order to find the optimal set of [hyperparameters](#hyperparamters) or optimal [model](#model).

<span id="cross-validation"><b><u>Cross-Validation</u></b></span>: Performing [holdout validation](#holdout-validation) multiple times for different [validation sets](#validation-set), then taking an average for the [model's](#model) [performace measures](#performance-measure) to get a more accurate depiction of how particular [models](#modelas) or particular sets of [hyperparameters](#hyperparameter) will perform. 

<span id="train-dev-set"><b><u>Train-Dev Set</u></b></span>: A subset of the [training set](#train-set) used to evaluate whether the [model](#model) is fit too well to the particular [training set](#train-set) (i.e. see [overfitting](#overfitting)).

<span id="supervised-learning-basics"><h3 style="text-align:center;">Supervised Learning Basics</h3></span>

<hr style="width:10%;">

<span id="supervised-learning"><b><u>Supervised Learning</u></b></span>: When the data that's fed to the [model](#model) during [training](#training) has a true, known value. For example, the data could be a set of animal images with the type of animal in the photo, or a set of variables for given houses along with the house price. In this case, the [training set](#train-set) typically consists of a [feature](#feature) matrix and a [label](#label) vector or a [target](#label) value. 

During [training](#training), it's typical for the [model's](#model) performance on the [training set](#train-set) to be evaluated by comparing the true [target](#target) to that predicted by the [model](#model). The [model parameters](#model-parameter) are then changed accordingly. After the [model](#model) is [trained](#trained), it can be fed a new piece of data (set of [features](#feature) without the [label](#label)) and it will predict the [label](#label) for that data. 

<span id="classification"><b><u>Classfication</u></b></span>: A type of [supervised learning](#supervised-learning) task in which the [label](#label) is some discrete value, for example a type of animal or a yes or no answer. In this case, the [label](#label) is typically a vector with a number of indices equal to the number of possible categories (e.g. for yes or no answers, the vector size is two; for types of animal the vector size could be huge depending on how many animals you want your model to choose from). For each data point, the [label](#label) has a 1 in the index corresponding to the correct category and a 0 for all other indices. For example, let's say we have a [model](#model) to predict whether the animal in a photo is a dog, a cat, an elephant, or a fish. The [labels](#labels) for each would look like:

$$
y_\text{dog} = 
\begin{pmatrix}
1 \\ 0 \\ 0 \\ 0 
\end{pmatrix}
\qquad
y_\text{cat} = 
\begin{pmatrix}
0 \\ 1 \\ 0 \\ 0 
\end{pmatrix}
\qquad
y_\text{elephant} = 
\begin{pmatrix}
0 \\ 0 \\ 1 \\ 0 
\end{pmatrix}
\qquad
y_\text{fish} = 
\begin{pmatrix}
0 \\ 0 \\ 0 \\ 1 
\end{pmatrix}
$$
where the first index corresponds to "dog", the second index to "cat", the third index to "elephant", and the fourth index to "fish". For this type of task, the [model](#model) would then predict which category a given data point belongs to. There are many different types of classification [algorithms](#algorithm) that we'll discuss in more detail below. 

<span id="regression"><b><u>Regression</u></b></span>: A type of [supervised learning](#supervised-learning) task in which the [target](#label) is some continuous value, for example the price of a house or the amount of energy consumed per hour. In this case, the [target](#target) is just some number and the [model](#ch1-model) predicts the value for a given data point. There are many different types of regression models that we'll discuss in more detail below.

<span id="linear-regression"><b><u>Linear Regression</u></b></span>: A [regression](#regression) [algogrithm](#algorithm) in which we try to fit the [training data](#train-set) to a line. For example, let's say we [train](#train) a linear [model](#model) to predict house prices from 100 data points each with two different [features](#feature) (e.g. each data point has a value for the (1) house size and the (2) number of rooms). Our [model](#model) would look something like:
$$
Y = X \Theta 
\qquad 
Y = 
\begin{pmatrix} 
y_1 \\ \vdots \\ y_N 
\end{pmatrix}
\qquad 
X = 
\begin{pmatrix} 
1 & x_{1,1} & x_{1,2} \\ 
1 & x_{2,1} & x_{2,2} \\ 
\vdots & & \vdots \\ 
1 & x_{100,1} & x_{100,2} 
\end{pmatrix} 
\qquad 
\Theta = 
\begin{pmatrix} \theta_0 \\ \theta_1 \\ \theta_2 
\end{pmatrix} 
$$
where $Y$ is the vector of [target](#label) values (house prices for each data point), $\Theta$ is the vector of [model parameters](#model-parameter) (one parameter for the intercept and two slope parameters for each of the features), and $X$ is the [feature](#feature) matrix (with an additional column of 1's for the intercept). We then get a set of equations:
$$
\begin{align}
y_1 &= \theta_0 + x_{1,1}\theta_1 + x_{1,2}\theta_2 \\
&\vdots \\
y_{100} &= \theta_0 + x_{100,1}\theta_1 + x_{100,2}\theta_2
\end{align}
$$
For our example, these equations are:
$$
\begin{align}
(\text{house price})_1 &= \theta_0 + (\text{house size})_1 \theta_1 + (\text{number of rooms})_1 \theta_2 \\
&\vdots \\
(\text{house price})_{100} &= \theta_0 + (\text{house size})_{100} \theta_1 + (\text{number of rooms})_{100} \theta_2
\end{align}
$$

The [model parameters](#model-parameters), $\Theta$, typically start as small, random values and then are updated some number of times throughout [training](#training) until the [training](#training) is ended. When the [model](#model) is then given some unlabeled data point, $X_\text{new}$, it predicts the [target](#target) value, $y_{pred}$, given the learned [model parameters](#model-parameter), $\Theta$ (i.e. if we give the [model](#model) a new (1) house size and (2) number of rooms, it will tell us the predicted house price):
$$
y_{pred} = \theta_0 + x_{\text{new},1}\theta_1 + x_{\text{new},2}\theta_2
$$
For our example, this is just:
$$
(\text{house price})_\text{pred} = \theta_0 + (\text{house size})_\text{new} \theta_1 + (\text{number of rooms})_\text{new} \theta_2
$$

<span id="unsupervised-learning-basics"><h3 style="text-align:center;">Unsupervised Learning Basics</h3></span>

<hr style="width:10%;">

<span id="unsupervised-learning"><b><u>Unsupervised Learning</u></b></span>: When the data that's fed to the [model](#model) during [training](#training) is just data without any true or known values attached to each point. For example, the data could be a set of images without any captions or a long list of customer [features](#features). Instead of [training](#training) a [model](#model) to predict a particular [target](#label), unsupervised learning typically finds unknown or unseen patterns in the data, such as grouping data into similar [clusters](#clustering) or [finding outliers](#anomaly-detection) in the data. The [training](#training) process is then significantly different compared to [supervised learning](#supervised-learning) since there's no true value to compare to predictions in order to evaluate the [model's](#model) performance. 

<span id="clustering"><b><u>Clustering</u></b></span>: A type of [unsupervised learning](#unsupervised-learning) task in which the data is broken up into similar groups. For example, if the data is a set of animal photos, a clustering model will learn patterns in the data and may group similar looking animals into groups based on learned [features](#features).

<span id="hierarchical-clustering"><b><u>Hierarchical Clustering</u></b></span>: A type of [unsupervised learning](#unsupervised-learning) task and a subset of [clustering](#clustering) in which groups can be nested in complex ways. 

<span id="visualization"><b><u>Visualization</u></b></span>: A type of [unsupervised learning](#unsupervised-learning) task that can be used to visualize how data is organized.

<span id="dimensional-reduction"><b><u>Dimensional Reduction</u></b></span>: A type of [unsupervised learning](#unsupervised-learning) task that can be used to reduce the number of [features](#features) for a given dataset. This is helpful for datasets with large amounts of different variables, some of which may be irrelevant and some of which can be combined together into more informative [features](#features). 

<span id="anomaly-detection"><b><u>Anomaly Detection</u></b></span>: A type of [unsupervised learning](#unsupervised-learning) task that can be used to flag unusual data instances, for example a suspicious purchase or a defective product. 

<span id="association-rule-learning"><b><u>Association Rule Learning</u></b></span>: A type of [unsupervised learning](#unsupervised-learning) task in which relations are found between data points through building rules based on if-then statements. 

<span id="other-supervised-learning"><h3 style="text-align:center;">Other Supervised Learning</h3></span>

<hr style="width:10%;">

<span id="semi-supervised-learning"><b><u>Semi-Supervised Learning</u></b></span>: When some of the data that's fed to the [model](#model) is [labeled](#label) and the rest is unlabeled. In this case, most of the data is typically unlabeled since it's easier to find and create. It can be time consuming and resource intensive to [label](#label) large amounts of data. 

<span id="self-supervised-learning"><b><u>Self-Supervised Learning</u></b></span>: When the data that's fed to the [model](#model) is inherently unlabled *but* treated in some way as to generate [labels](#label). For example, let's say we have a huge amount of text data. This is inherently unlabeled, it's just a bunch of text without any "true" values associated with it and trying to manually [label](#label) it in some way is untenable. But, what if we mask some of the words in the text then [train](#training) the [model](#model) to predict the words that are masked? Well, now we have the true values for masked words that we can then compare to the [model's](#model) predictions, which is essentially a [supervised learning](#supervised-learning) task. The data that's fed to the model during [training](#training) is essentially [labeled](#label) because we have text with masks and [labels](#label) corresponding to what those masks truly are. 

Self-supervised learning is then a technique that benefits from both the scale of the mass amounts of unlabeled data that is typically available for [unsupervised learning](#unsupervised-learning) tasks and the ease of true comparisons for [supervised learning](#supervised-learning) tasks.

<span id="reinforcement-learning-basics"><h3 style="text-align:center;">Reinforcement Learning Basics</h3></span>

<hr style="width:10%;">

<span id="reinforcement-learning"><b><u>Reinforcement Learning</u></b></span>: When the data that's fed to the [agent](#agent) is a space of potential actions that can be performed and the [agent](#agent) learns the best [policy](#policy) to take based on a reward system.

<span id="agent"><b><u>Agent</u></b></span>: In [reinforcement learning](#reinforcement-learning), an agent is a system that can take an action from some pre-defined actions. It's then rewarded based on this action and learns the best actions to take in order to maximize its total reward over time. 

<span id="policy"><b><u>Policy</u></b></span>: In [reinforcement learning](#reinforcement-learning), the [agent](#agent) learns an action policy - what action to take in any given situation - based on the reward system.

<span id="batch-learning-basics"><h3 style="text-align:center;">Batch Learning Basics</h3></span>

<hr style="width:10%;">

<span id="batch-learning"><b><u>Batch Learning</u></b></span>: (*AKA Offline Learning*) When [training](#training) uses the full [training set](#train-set) in one go. No breaking the data into parts and training little by little, just feeding the entire dataset all at once, as once entire *batch*. Any time we want to retrain the [model](#model) in light of some new data, we would have to feed the entire dataset (plus the new data), every time. 

<span id="data-drift"><b><u>Data Drift</u></b></span>: (*AKA Model Rot*) When the performance of a [model](#model) degrades due to its static nature. When we [train](#train) a [model](#model), it learns the optimal parameters and that's it. It's frozen in time. It doesn't continue learning unless we continue [training](#training) it. As the world changes, its data changes with it and at some point the data may be so different from the [training data](#train-set) that was fed to the [model](#model) that the [model](#model) begins to perform poorly. 

<span id="online-learning-basics"><h3 style="text-align:center;">Online Learning Basics</h3></span>

<hr style="width:10%;">

<span id="online-learning"><b><u>Online Learning</u></b></span>: (*AKA Incremental Learning, [Mini-Batch](#mini-batch) Learning*) When [training](#training) utilizes groups of the [training set](#train-set) instead of the full [batch](#batch-learning). In this case, the data can be broken up as small as individual data points, or as larger groups called *[mini-batches](#mini-batch)*. The [model](#model) is then [trained](#training) on these groups one-by-one and updates its [parameters](#model-parameter) accordingly. After all the groups have been fed to the [model](#model), it's again frozen in time. But, it's now much easier to feed new data (compared to [batch learning](#batch-learning)) because we can break the new data up into small groups and [train](#training) the [model](#model) incrementally. 

<span id="mini-batch"><b><u>Mini-Batch</u></b></span>: (*AKA Batch*) A group of the full [training set](#train-set) that the [algorithm](#algorithm) [trains](#training) on during [online learning](#online-learning). In this case, the [training data](#train-set) is typically divided into some number of mini-batches and the [algorithm](#algorithm) [trains](#training) on each mini-batch until it has seen all of the data. 

The fact that a *mini-batch* is also called commonly called a *batch* for short makes the name for [batch learning](#batch-learning) confusing. In [batch learning](#batch-learning), the *batch* is actually the entire [training set](#train-set).

<span id="out-of-core-learning"><b><u>Out-of-Core Learning</u></b></span>: Breaking a large dataset up into small groups that can then be fed to a system, group by group, through [incremental learning](#online-learning) in order to sidestep memory limitations.

<span id="loss-function"><b><u>Loss Function</u></b></span>: (*AKA Cost Function*) A measure of how far a [model's](#model) predictions are from the true [labels](#label). This is typically calculated as some sort of difference between the two for each data point in the [training set](#train-set) (or [mini-batch](#mini-batch)). When [training](#training), it's typical to try to find the [model parameters](#model-parameters) which minimize the loss.

<span id="mse"><b><u>Mean Squared Error (MSE)</u></b></span>: A popular [loss function](#loss-function) used in [linear regression](#linear-regression). 
$$
\text{MSE} = \frac{1}{N}\sum_{i=1}^N \left(y_i - y_\text{pred}\right)^2
$$
where $y_\text{pred}$ is the predicted value and $y_i$ is the [label](#label) for the $i$-th data point out of $N$ total data points. For a [linear regression](#linear-regression) [model](#model) with two [features](#feature), the MSE is:
$$
\text{MSE} = \frac{1}{N}\sum_{i=1}^N \left(y_i - \theta_0 - \theta_1 x_{i,1} - \theta_2 x_{i,2}\right)^2
$$

By squaring the difference between the predicted and actual values, we emphasize errors that are really small (by making them even smaller) and really large (by making them even larger). By dividing by $N$, we scale the calculation to that of one data point. Notice that the units are of the error squared rather than the error.

<span id="rmse"><b><u>Root Mean Squared Error (RMSE)</u></b></span>: A popular [loss function](#loss-function) used in [linear regression](#linear-regression). 
$$
\text{RMSE} = \sqrt{\frac{1}{N}\sum_{i=1}^N \left(y_i - y_\text{pred}\right)^2}
$$
where $y_\text{pred}$ is the predicted value and $y_i$ is the [label](#label) for the $i$-th data point out of $N$ total data points. For a [linear regression](#linear-regression) [algorithm](#algorithm) with two [features](#feature), the RMSE is:
$$
\text{RMSE} = \sqrt{\frac{1}{N}\sum_{i=1}^N \left(y_i - \theta_0 - \theta_1 x_{i,1} - \theta_2 x_{i,2}\right)^2}
$$

By squaring the difference between the predicted and actual values, we emphasize errors that are really small (by making them even smaller) and really large (by making them even larger). By dividing by $N$, we scale the calculation to that of one data point, and by taking the square-root, we scale the calculation to the same units as the error (instead of the error squared).

<span id="gradient-descent"><b><u>Gradient Descent</u></b></span>: A method for finding the [model parameters](#model-parameters) that minimizes the [loss](#loss-function) during [training](#training). A broad overview of the process is: (1) calculate the [loss](#loss-function) for the given [mini-batch](#mini-batch), (2) calculate the derivate of the [loss](#loss-function) with respect to each of the [model parameters](#model-parameter) (i.e. *minimize* the [loss](#loss-function) by finding the *gradient*), (3) update the [model parameters](#model-parameter) based on some small factor called the [learning rate](#learning-rate) (a [hyperparameter](#hyperparameter) of the gradient descent algorithm), (4) repeat steps (1)-(3) until [training](#training) is ended. 

Take the [linear regression](#linear-regression) [algorithm](#algorithm) as an example. Let's say we have a [training set](#train-set) with two [features](#features) and we want to use the [MSE](#mse) as the [loss function](#loss-function). Taking a [mini-batch](#mini-batch) size of $N$, the [training](#training) process using gradient descent would look something like:
1. Calculate the [loss](#loss-function) for the [mini-batch](#mini-batch):
$$
\text{MSE} = \frac{1}{N}\sum_{i=1}^N \left(y_i - \theta_0 - \theta_1 x_{i,1} - \theta_2 x_{i,2}\right)^2
$$
2. Calculate the derivate of the [loss](#loss-function) with respect to each of the [model parameters](#model-parameter):
$$
\begin{align}
\frac{\partial{\text{MSE}}}{\partial{\theta_0}} &= -\frac{2}{N}\sum_{i=1}^N \left(y_i - \theta_0 - \theta_1 x_{i,1} - \theta_2 x_{i,2}\right) \\
\frac{\partial{\text{MSE}}}{\partial{\theta_1}} &= -\frac{2}{N}\sum_{i=1}^N x_{i,1}\left(y_i - \theta_0 - \theta_1 x_{i,1} - \theta_2 x_{i,2}\right) \\
\frac{\partial{\text{MSE}}}{\partial{\theta_2}} &= -\frac{2}{N}\sum_{i=1}^N x_{i,2}\left(y_i - \theta_0 - \theta_1 x_{i,1} - \theta_2 x_{i,2}\right)
\end{align}
$$
3. Update the [model parameters](#model-parameter) based on the [learning rate](#learning-rate):
$$
\begin{align}
\theta_0 &= \theta_0 - \alpha \frac{\partial{\text{MSE}}}{\partial{\theta_0}} = \theta_0 + \frac{2\alpha}{N}\sum_{i=1}^N \left(y_i - \theta_0 - \theta_1 x_{i,1} - \theta_2 x_{i,2}\right) \\
\theta_1 &= \theta_1 - \alpha \frac{\partial{\text{MSE}}}{\partial{\theta_1}} = \theta_1 + \frac{2\alpha}{N}\sum_{i=1}^N x_{i,1}\left(y_i - \theta_0 - \theta_1 x_{i,1} - \theta_2 x_{i,2}\right) \\
\theta_2 &= \theta_2 - \alpha \frac{\partial{\text{MSE}}}{\partial{\theta_2}} = \theta_2 + \frac{2\alpha}{N}\sum_{i=1}^N x_{i,2}\left(y_i - \theta_0 - \theta_1 x_{i,1} - \theta_2 x_{i,2}\right)
\end{align}
$$
4. Repeat steps (1)-(3) for all [mini-batches](#mini-batch).

In the end, we've hopefully found an optimal set of [model parameters](#model-parameters) by continuously moving in the direction of the minimum loss.

<span id="learning-rate"><b><u>Learning Rate</u></b></span>: The [hyperparameter](#hyperparameter) in [gradient descent](#gradient-descent) which affects how much the [model parameters](#model-parameters) are updated during [training](#training). We typically use the variable $\alpha$ to denote the learning rate.

If the learning rate is too small, each update to the [model parameters](#model-parameter) is very a small change and it takes too long for the [model](#model) to learn. In this case, it will probably not take into account outliers or new data very well. If the learning rate is too large, however, the [model parameters](#model-parameter) jump widly with each update and the [model](#model) may fall into suboptimal local minima or favor the newest data. In this case, it will probably [test](#testing) well on the most recent data, but [generalize](#generalization) poorly to both previously seen data or unseen data. This is something called [catastrophic forgetting](#catastrophic-forgetting).

<span id="catastrophic-forgetting"><b><u>Catastrophic Forgetting</u></b></span>: (*AKA Catastrophic Interference*) A performance degradation that happens when the [learning rate](#learning-rate) for a system is too large. In this case, the [model](#model) favors the most recently seen data and seems to *forget* ([tests](#testing) poorly on) data that it's already seen.

<span id="ml-challenges"><h1 style="text-align:center;">ML Challenges</h1></span>

<hr style="width:50%;">

<span id="unreasonable-effectiveness-of-data"><b><u>Unreasonable Effectiveness of Data</u></b></span>:

<span id="sampling-noise"><b><u>Sampling Noise</u></b></span>: When the data is not [representative](#representative) of real-world data because the sample size is too small (and hence random noise is not averaged out). 

<span id="sampling-bias"><b><u>Sampling Bias</u></b></span>: When the data is not [representative](#representative) of real-world data because of inherent flaws in how the data is obtained. For example, let's say we survey students at a university. If we want to then [test](#testing) on data that will include all types of people (not just university students), our data will be inherently biased and our [model](#model) will most likely perform poorly.

<span id="overfitting"><b><u>Overfitting</u></b></span>: When [training](#training) results in a [model](#model) that takes into account the idosyncracies of the [training data](#train-set) too much so that it [generalizes](#generalizes) poorly to unseen data. In this case, the [model](#model) will perform extremely well when evaluated on the [training data](#train-set) but will perform poorly when evaluated on the [test set](#test-set) (as well as the [validation set](#validation-set) and, most telling, the [train-dev set](#train-dev-set). Typically, this means that the [model](#model) is too complex relative to the [training data](#train-set) size.

<span id="regularization"><b><u>Regularization</u></b></span>: A method to inhibit [overfitting](#overfitting) by constraining the [model parameters](#model-parameter) is some way and effectively reducing the number of [degrees of freedom](#degrees-of-freedom).

<span id="underfitting"><b><u>Underfitting</u></b></span>: When [training](#training) results in a [model](#model) that can't take into account the complexity of the data. In this case, the [model](#model) typically performs poorly on all data sets (i.e. [training](#train-set), [testing](#test-set), [validation](#validation-set), [train-dev set](#train-dev-set)) and needs more [model parameters](#model-parameters).
